# 4.Data Preparation

```python
weather_Data.drop(labels=["Unnamed: 0","Temp C"],axis = 1,inplace = True)
del weather_Data["Ozone"]
```

```python
weather_Data.head(10)
```

```python
weather_Data[weather_Data["Solar.R"].isna()]
```

```python
weather_Data.groupby("Weather")["Solar.R"].mean()
```

```python
weather_Data["Solar.R"] = weather_Data.groupby("Weather")["Solar.R"].transform(lambda x:x.fillna(x.mean().round()))
weather_Data
```

```python
weather_Data.isna().sum()
```

```python
weather_Data.dropna(inplace=True)
```

```python
weather_Data.isna().sum()
```

```python
weather_Data.dtypes
```

```python
weather_Data.Month.unique()
```

```python
weather_Data["Month"] = weather_Data["Month"].replace( to_replace="May",
    value=5)
weather_Data["Month"] = weather_Data["Month"].astype("int")
weather_Data.dtypes
```

```python
from sklearn.preprocessing import LabelEncoder
LE = LabelEncoder()
weather_Data_Copy1 = weather_Data.copy()
weather_Data_Copy1
```

```python
weather_Data_Copy1["Weather"] = LE.fit_transform(weather_Data_Copy1["Weather"])
weather_Data_Copy1.dtypes
weather_Data_Copy1
```

```python
from sklearn.preprocessing import OneHotEncoder

from sklearn.preprocessing import OneHotEncoder
import pandas as pd

# Copy dataframe
weather_Data_Copy3 = weather_Data.copy()

# Create encoder
ohe = OneHotEncoder(sparse_output=False)

# Encode Weather column
encoded = ohe.fit_transform(weather_Data_Copy3[['Weather']])

# Convert encoded data to dataframe
encoded_df = pd.DataFrame(
    encoded,
    columns=ohe.get_feature_names_out(['Weather']),
    index=weather_Data_Copy3.index
)

# Drop original Weather column
weather_Data_Copy3 = weather_Data_Copy3.drop('Weather', axis=1)

# Combine both dataframes
weather_Data_Copy3_fin = pd.concat(
    [weather_Data_Copy3, encoded_df],
    axis=1
)

print(weather_Data_Copy3_fin)
```
